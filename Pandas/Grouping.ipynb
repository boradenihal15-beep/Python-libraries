```python
import pandas as pd
import numpy as np
```

# Grouping

* Grouping is used to group the data in categories.

* it makes easy to perform  aggregate methods on categories.

* e.g calculating highest marks of student from each dept.


```python
movies=pd.read_csv('imdb-top-1000.csv')
dev=pd.read_csv('deliveries.csv')
```

```python
#groupby

movies.groupby('Genre').max()['IMDB_Rating']#applied agg method max to get max rating for each genre
```

```python
# find the top 3 genres by total earning

movies.groupby('Genre').sum()['Gross'].sort_values(ascending=False).head(3)
```

```python
# find the genre with highest avg IMDB rating
movies.groupby("Genre")['IMDB_Rating'].mean().sort_values(ascending=False).head(1)
```

```python
# find director with most popularity
movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
# find the highest rated movie of each genre
movies.groupby(['Genre'])['IMDB_Rating'].max()
```

```python
# find number of movies done by each actor
movies.groupby('Star1')['Series_Title'].count().sort_values(ascending=False)
```

```python
#groupby attributes and methods

#len
print(len(movies.groupby("Genre"))) #Total number of groups

#size
movies.groupby("Genre").size() #Total items in each group

#first()/last()/nth() value
genre=movies.groupby('Genre')
genre.first() #first value of each genre
genre.last() #last value of each genre
genre.nth(6) #6th movie of each  genre

```

```python
#fetch a group

genre.get_group('Action')
#or
movies[movies['Genre']=='Action']
```

```python
#get groups and values

genre.groups
```

```python
#descibe

#stats of each genre
genre.describe()
```

```python
#sample

#random value from each genre
genre.sample(2,replace=True) #if want custom no of values
```

```python
#nunique

genre.nunique()
```

```python
#agg
genre.sum()
# genre.mean()

#This method of applying aggregate method will give agg for all columns with different output
#but what if we want to apply diff agg to diff column
#then we use agg using dict

genre.agg({
    'Runtime':'mean',
    'IMDB_Rating':'sum',
    'Metascore':'max'
})
#so we can apply any agg to any column

#agg using list
#if we want to apply multiple agg to each column we pass list of agg methods

genre.agg(['sum','max','min']) #now we can get all three agg to each colmn

genre[['Runtime','IMDB_Rating','Metascore']].agg(['mean','sum','min','max'])

#also we can combine both methods
genre.agg({
    'Runtime':['mean','sum'],
    'IMDB_Rating':['max','min'],
    'Metascore':['max','mean']
})
```

```python
#Looping
df=[]
for group,data in genre:
  max_rating=data[data['IMDB_Rating']==data['IMDB_Rating'].max()]
  df.append(max_rating[['Genre','Series_Title']])

df=pd.concat(df)
df
```

```python
#apply
#split apply combine
#we can apply buildin or custom methods on groups

# find number of movies starting with A for each group
def foo(group):
  return group['Series_Title'].str.startswith("A").sum()

genre.apply(foo)
```
