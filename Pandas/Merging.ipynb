```python
import pandas as pd
import numpy as np
```

```python
courses = pd.read_csv('courses.csv')
students = pd.read_csv('students.csv')
nov = pd.read_csv('reg-month1.csv')
dec = pd.read_csv('reg-month2.csv')

matches = pd.read_csv('matches.csv')
delivery = pd.read_csv('deliveries.csv')
```

```python
#concat
#concat is used to concat two dataframes vertically
#the data of two dataframes at stacked vertically

pd.concat([nov,dec])

#ignoreindex
# after every stack the index start new so to continue indexes we use ignoreindex

pd.concat([nov,dec],ignore_index=True)

#multiindex
#if you want to sep the stacked dataframes with custom index we use keys
#after using keys a 2nd index column is created with default index column that's why it is called as multiindex

pd.concat([nov,dec],keys=['nov','dec']) #but we cant us ignoreindex

# now we can locate data based on custom index
regs=pd.concat([nov,dec],keys=['nov','dec'])
regs.loc['nov']

regs.loc[('nov',0)] #gives a specific row from specific stack 1st value of nov
regs.loc[('nov',3)] #gives a specific row from specific stack 3rd value of nov

#horizontal stacking
pd.concat([nov,dec],axis=1)
# in this it is not necassery that rows should be equal it will assing NAN values to rows
#as there a more rows in dec the NAN is assign to rows of nov
```

```python
# innerjoin

#All common values from both tables
students.merge(regs,how='inner',on='student_id')
```

```python
#left join
# common and left table values
#NAN if regs values are not in students
students.merge(regs,how='left',on='student_id')
```

```python
#right join
#comon and right table values
students.merge(regs,how='right',on='student_id')
#NAN values are student_id 45 50 38 bcoz it has no students
```

```python
#outer join
students.merge(regs,how='outer',on='student_id')
#all values from both tables
```
