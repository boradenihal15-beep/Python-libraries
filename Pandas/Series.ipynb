# **Series**

Import pandas

```python
import numpy as np
import pandas as pd
```

```python
#string
string=['hi','bye','why']
series=pd.Series(string)
series
```

```python
#intergers
intergers=[1,3,2,4,5]
series=pd.Series(intergers)
series
```

```python
#custom index
names=['karan','varan']
marks=[90,60]

students=pd.Series(data=marks,index=names)

students
```

```python
#Name to series
names=['karan','varan']
marks=[90,60]

students=pd.Series(data=marks,index=names,name='student_marks')

students
```

```python
#Series of dict
marks={
    'Nihal':90,
    'karan':40,
    'Nihal':30
}
students=pd.Series(marks)
students
```

```python
#Attributes of series

#size
students.size

#dtype
students.dtype

#index
students.index

#values
students.values

#isunique
students.is_unique
```

```python
#Reading the dataset

#one column
data=pd.read_csv("subs.csv")
subs=data.squeeze()
subs

#Two column
data=pd.read_csv('kohli_ipl.csv')
vk=data['runs'].squeeze()

data=pd.read_csv('bollywood.csv')
movies=data['lead'].squeeze()
```

```python
#head and tail
movies.head() #first default 5
movies.tail() #last default 5
movies.head(15) #first 15

movies.sample() #random value
movies.sample(3) #random 3 value
```

```python
#value count
movies.value_counts() #total movies of actress (total values of index)
```

```python
#sort values
vk.sort_values(ascending=False).head(1).values[0]
```

```python
#sort by index
data=pd.read_csv('kohli_ipl.csv')
vk=pd.Series(data=data['runs'],index=data['match_no'])
vk.sort_index()
```

```python
#maths methods

#count
vk.count()

#sum
subs.sum()

#product
subs.product()

# mean -> median -> mode -> std -> var
subs.mean()
print(vk.median())
print(movies.mode())
print(subs.std())
print(vk.var())
```

```python
# min/max
subs.min()
subs.max()
```

```python
#descibe
vk.describe()
```

```python
#indexing

#pos indexing
subs[1]

#Neg indexing
# subs[-1] #gives error
students[-1] #works only for custom indexes

#string index
students['Nihal']
```

```python
#Slicing
vk[1:6]
vk[:6:2]
vk[-5::]
vk[::-1]
```

```python
#fancy indexing
subs[[1,3,4]]
```

```python
#Editing series

students[1]=40
students['Nihal']=40
students
students['suraj']=100
students

#by slicing
students[1:2]=100
students

#by fancy indexing
students[[0,2]]=100
students
```

```python
# len/type/dir/sorted/max/min
print(len(subs))
print(type(subs))
print(dir(subs))
print(sorted(subs))
print(min(subs))
print(max(subs))
```

```python
#type conversion
list(students)
dict(students)
```

```python
#Membership operators

'Nihal' in students #it is based on index
100 in students.values #to check in values
```

```python
#looping

#looping is based on values
for i in students:
  print(i)

for i in students.index: #to do on indexes
  print(i)
```

```python
#Arithmatic operator
100+students #also called as broadcasting one values is added to all rows
```

```python
#relational operator
vk[vk>100]
```

```python
#Boolean index

#No of 50's
vk[vk>50].count()

#Actor more than 20 movies
count=movies.value_counts()
count[count>20]
```
